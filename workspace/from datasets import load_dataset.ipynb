## Download the dataset

This notebook loads the Hugging Face dataset and saves it to an Excel file at `data/emotion/emotion.xlsx`.
- Sheets: `train`, `validation`, `test`


```python
from datasets import load_dataset

ds = load_dataset("dair-ai/emotion")
ds
```

```python
import os
import pandas as pd

out_dir = os.path.join("data", "emotion")
os.makedirs(out_dir, exist_ok=True)

xlsx_path = os.path.join(out_dir, "emotion.xlsx")

# Excel export needs an engine like `openpyxl`.
try:
    import openpyxl  # noqa: F401
except ModuleNotFoundError as e:
    raise ModuleNotFoundError(
        "To export to Excel, install openpyxl first: %pip install openpyxl"
    ) from e

with pd.ExcelWriter(xlsx_path, engine="openpyxl") as writer:
    for split, dset in ds.items():
        dset.to_pandas().to_excel(writer, sheet_name=split, index=False)

xlsx_path
```
